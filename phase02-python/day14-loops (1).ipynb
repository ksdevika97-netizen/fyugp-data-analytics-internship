```python
scores = [78, 45, 67, 89, 34, 56, 92, 48, 61, 39]

for score in scores:
    if score >= 50:
        print(score, "Pass")
    else:
        print(score, "Fail")

passed=0
failed=0

for score in scores:
    if score>=50:
        passed=passed+1
    else:
        failed=failed+1

print(f"{passed} Students passed")
print(f"{failed} Students failed")
```
