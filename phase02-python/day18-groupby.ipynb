```python
import pandas as pd
df=pd.read_csv("SALE.csv")
df
```

```python
#Total Profit by Category
df.groupby("Category")["Profit"].sum()
```

```python
#Average sales by category
df.groupby("Category")["Sales"].mean()
```

```python
# Count of rows by Category
df["Category"].value_counts()
```
